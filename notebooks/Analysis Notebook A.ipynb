```python
# Who receives money in a loan repayment? That's the lender.
# Now find what transaction type sends money TO customers from that same entity.
from pyspark.sql.functions import col
repayment_types = [
    'Loan Principal Collection via API',
    'Loan Interest via API',
    'Loan Payment Via API',
    'Loan Penalty via API'
    ]

# Who is the CREDIT party (lender) in repayments?
df.filter(col('TRANSACTION_TYPE').isin(repayment_types)).groupBy('CREDIT_PARTY_TYPE', 'CREDIT_ACCOUNT_TYPE').count().orderBy('count', ascending=False).display(20, truncate=False)
```

```python
# Get the most common CREDIT_PARTY_ID in repayments (the lender)
lender_id = df.filter(col('TRANSACTION_TYPE').isin(repayment_types)).groupBy('CREDIT_PARTY_ID').count().orderBy('count', ascending=False).first()['CREDIT_PARTY_ID']

print("Lender ID:", lender_id)

# Now find what this lender sends OUT (disbursements)
df.filter(col('DEBIT_PARTY_ID') == lender_id).groupBy('TRANSACTION_TYPE').count().orderBy('count', ascending=False).display(20, truncate=False)
```

```python
df.filter(col('DEBIT_PARTY_TYPE') == 'Customer').select('DEBIT_PARTY_ID', 'DEBIT_ACCOUNT_TYPE').distinct().groupBy('DEBIT_ACCOUNT_TYPE').count().display(truncate=False)
```

```python
# How many unique borrowers and what's the default rate
from pyspark.sql.functions import col, countDistinct
lender_id = 'E7C89F8C4A27F173'

# Unique customers who received a loan
borrowers = df.filter(
    (col('TRANSACTION_TYPE') == 'Loan Payment Via API') &
    (col('DEBIT_PARTY_ID') == lender_id) &
    (col('CREDIT_PARTY_TYPE') == 'Customer')
).select(col('CREDIT_PARTY_ID').alias('user_id')).distinct()

print("Total borrowers:", borrowers.count())

# Customers who have a penalty transaction (proxy for bad outcome)
penalised = df.filter(
    (col('TRANSACTION_TYPE') == 'Loan Penalty via API') &
    (col('DEBIT_PARTY_TYPE') == 'Customer')
).select(col('DEBIT_PARTY_ID').alias('user_id')).distinct()

print("Borrowers with penalties:", penalised.count())

# Overlap: borrowers who also have penalties
defaulters = borrowers.join(penalised, on='user_id', how='inner')
print("Borrowers + penalties (overlap):", defaulters.count())
```

```python
# Borrowers with zero repayments (strict defaulters)
# Customers who made any repayment (principal or interest)
repayers = df.filter(
    col('TRANSACTION_TYPE').isin([
        'Loan Principal Collection via API',
        'Loan Interest via API'
    ]) &
    (col('DEBIT_PARTY_TYPE') == 'Customer')
).select(col('DEBIT_PARTY_ID').alias('user_id')).distinct()

print("Borrowers who made at least one repayment:", repayers.count())

# Borrowers who never repaid
never_repaid = borrowers.join(repayers, on='user_id', how='left_anti')
print("Borrowers with zero repayments:", never_repaid.count())
```

```python
import subprocess, sys, os

# Point subprocess at the real-data CSVs written by build_pipeline (underscore path).
# SEQCREDIT_DATA_DIR makes the subprocess's DATA_DIR match, so results also save there.
# Pop SEQCREDIT_EPHEMERAL so the benchmark actually writes cv_results_*.csv to disk.
env = {**os.environ}
env["SEQCREDIT_DATA_DIR"]           = "/tmp/seqcredit_model"
env["SEQCREDIT_USER_FEATURES_FILE"] = "/tmp/seqcredit_model/user_features.csv"
env["SEQCREDIT_USER_LABELS_FILE"]   = "/tmp/seqcredit_model/user_labels.csv"
env.pop("SEQCREDIT_EPHEMERAL", None)

result = subprocess.run(
    [sys.executable, "src/seqcredit_model/run_cv_benchmark.py"],
    capture_output=True, text=True,
    cwd="/tmp/seqcredit-model",
    env=env,
)

print(result.stdout[-6000:])
if result.returncode != 0:
    print("STDERR:", result.stderr[-3000:])
```
